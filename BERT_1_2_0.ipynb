```python
# fixando semente
import os
import random
import re
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_curve, auc
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.colors import LinearSegmentedColormap
from transformers import BertTokenizer
from datasets import load_dataset
from lime.lime_text import LimeTextExplainer

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False
    os.environ['PYTHONHASHSEED'] = str(seed)
    print(f"[OK] semente fixada em: {seed}")

set_seed(42)
```

```python
print("Baixando o Dataset ")
dataset_hf = load_dataset("mrjunos/depression-reddit-cleaned", trust_remote_code=True)
df = pd.DataFrame(dataset_hf['train'])
df = df.rename(columns={'text': 'texto_limpo'})

def limpar_texto_redes_sociais(texto):
    if not isinstance(texto, str):
        return ""
    texto = texto.lower()
    texto = re.sub(r'http\S+|www\S+|https\S+', '', texto, flags=re.MULTILINE)
    texto = re.sub(r'\@\w+', '', texto)
    texto = re.sub(r'[^\w\s]', '', texto)
    texto = re.sub(r'\s+', ' ', texto).strip()
    return texto

print("limpando")
df['texto_limpo'] = df['texto_limpo'].apply(limpar_texto_redes_sociais)

qtd_antes = len(df)
df = df[df['texto_limpo'].str.len() > 5].copy()
df = df.drop_duplicates(subset=['texto_limpo']).reset_index(drop=True)
qtd_depois = len(df)
print(f"{qtd_antes - qtd_depois} textos duplicados/vazios excluidos")
print(f"total final de registros únicos e limpos: {qtd_depois}")
```

```python
# divisão
df_treino, df_teste = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    stratify=df['label']
)
print(f"Treino: {len(df_treino)} | Teste: {len(df_teste)}")

print("tokenizando textos com o BertTokenizer")
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
train_encodings = tokenizer(df_treino['texto_limpo'].tolist(), truncation=True, padding='max_length', max_length=128)
test_encodings = tokenizer(df_teste['texto_limpo'].tolist(), truncation=True, padding='max_length', max_length=128)
print("tokenização concluída")
```

```python
#organização
class RedditDepressionDataset(Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {key: torch.tensor(val[idx]) for key, val in self.encodings.items()}
        item['labels'] = torch.tensor(self.labels[idx], dtype=torch.long)
        return item

    def __len__(self):
        return len(self.labels)

print("criando os objetos")
train_dataset = RedditDepressionDataset(train_encodings, df_treino['label'].tolist())
test_dataset = RedditDepressionDataset(test_encodings, df_teste['label'].tolist())

print("empacotando os dados em lotes (Batch Size = 32)")
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)
print(f"Treino: {len(train_loader)} lotes | Teste: {len(test_loader)} lotes.\n")

print("Vendo o Primeiro Lote ")
primeiro_lote = next(iter(train_loader))
print("Chaves contidas no lote: ", primeiro_lote.keys())
print("Dimensão dos input_ids: ", primeiro_lote['input_ids'].shape)
print("Dimensão da attention_mask: ", primeiro_lote['attention_mask'].shape)
print("Dimensão dos rótulos: ", primeiro_lote['labels'].shape)
print("\nRótulos das 32 amostras deste lote (0 = Não Depressivo, 1 = Depressivo):")
print(primeiro_lote['labels'])
```

```python
# arquitetura bert
import torch
import torch.nn as nn
import numpy as np

class BERTEmbeddings(nn.Module):
    def __init__(self, vocab_size=30522, hidden_size=768, max_len=512):
        super().__init__()
        self.word_embeddings = nn.Embedding(vocab_size, hidden_size)
        self.position_embeddings = nn.Embedding(max_len, hidden_size)
        self.segment_embeddings = nn.Embedding(2, hidden_size)
        self.layer_norm = nn.LayerNorm(hidden_size)
        self.dropout = nn.Dropout(0.1)

    def forward(self, input_ids, token_type_ids):
        seq_len = input_ids.size(1)
        position_ids = torch.arange(seq_len, dtype=torch.long, device=input_ids.device).unsqueeze(0)
        embeddings = self.word_embeddings(input_ids) + self.position_embeddings(position_ids) + self.segment_embeddings(token_type_ids)
        return self.dropout(self.layer_norm(embeddings))

class SelfAttention(nn.Module):
    def __init__(self, hidden_size=768, num_heads=12):
        super().__init__()
        self.num_heads = num_heads
        self.head_dim = hidden_size // num_heads
        self.query = nn.Linear(hidden_size, hidden_size)
        self.key = nn.Linear(hidden_size, hidden_size)
        self.value = nn.Linear(hidden_size, hidden_size)
        self.fc_out = nn.Linear(hidden_size, hidden_size)

    def forward(self, x, mask=None):
        N, length, _ = x.shape
        Q = self.query(x).reshape(N, length, self.num_heads, self.head_dim).transpose(1, 2)
        K = self.key(x).reshape(N, length, self.num_heads, self.head_dim).transpose(1, 2)
        V = self.value(x).reshape(N, length, self.num_heads, self.head_dim).transpose(1, 2)

        energy = torch.matmul(Q, K.transpose(-2, -1)) / np.sqrt(self.head_dim)
        if mask is not None:
            energy = energy.masked_fill(mask.unsqueeze(1).unsqueeze(2) == 0, -1e9)

        attention = torch.softmax(energy, dim=-1)
        out = torch.matmul(attention, V).transpose(1, 2).reshape(N, length, self.num_heads * self.head_dim)
        return self.fc_out(out)

class TransformerBlock(nn.Module):
    def __init__(self, hidden_size=768, num_heads=12):
        super().__init__()
        self.attention = SelfAttention(hidden_size, num_heads)
        self.norm1 = nn.LayerNorm(hidden_size)
        self.norm2 = nn.LayerNorm(hidden_size)
        self.feed_forward = nn.Sequential(
            nn.Linear(hidden_size, 3072),
            nn.GELU(),
            nn.Linear(3072, hidden_size)
        )
        self.dropout = nn.Dropout(0.1)

    def forward(self, x, mask=None):
        x = self.norm1(x + self.dropout(self.attention(x, mask)))
        return self.norm2(x + self.dropout(self.feed_forward(x)))

class BERT(nn.Module):
    def __init__(self, vocab_size=30522, hidden_size=768, num_layers=12):
        super().__init__()
        self.embeddings = BERTEmbeddings(vocab_size, hidden_size)
        self.encoder = nn.ModuleList([TransformerBlock(hidden_size, 12) for _ in range(num_layers)])

    def forward(self, input_ids, token_type_ids, mask=None):
        x = self.embeddings(input_ids, token_type_ids)
        for layer in self.encoder:
            x = layer(x, mask)
        return x

class BERTClassificadorDoZero(nn.Module):
    def __init__(self, hidden_size=768, num_classes=2):
        super().__init__()
        self.bert = BERT()
        self.classificador = nn.Linear(hidden_size, num_classes)

    def forward(self, input_ids, attention_mask):
        segment_ids = torch.zeros_like(input_ids)
        saida_bert = self.bert(input_ids, segment_ids, mask=attention_mask)
        return self.classificador(saida_bert[:, 0, :])

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"instanciando o BERT no dispositivo: {device}")
meu_modelo = BERTClassificadorDoZero().to(device)
```

```python
# treinamento
criterio_perda = nn.CrossEntropyLoss()
otimizador = AdamW(meu_modelo.parameters(), lr=2e-5)
epochs = 3

print("iniciando o treinamento")
for epoch in range(epochs):
    meu_modelo.train()
    total_loss = 0

    for lote_idx, batch in enumerate(train_loader):
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels = batch['labels'].to(device)

        otimizador.zero_grad()
        previsoes = meu_modelo(input_ids, attention_mask)
        loss = criterio_perda(previsoes, labels)

        total_loss += loss.item()
        loss.backward()
        otimizador.step()

        if lote_idx % 50 == 0:
            print(f"Época {epoch + 1}/{epochs} | Lote {lote_idx}/{len(train_loader)} | Loss: {loss.item():.4f}")

    print(f"--- Fim da Época {epoch + 1} | Perda Média: {total_loss / len(train_loader):.4f} ---")

print("Treinamento  concluído")
```

```python
# avaliação
meu_modelo.eval()
previsoes_finais, rotulos_reais = [], []

with torch.no_grad():
    for batch in test_loader:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels = batch['labels'].to(device)

        logits = meu_modelo(input_ids, attention_mask)
        preds = torch.argmax(logits, dim=1)

        previsoes_finais.extend(preds.cpu().numpy())
        rotulos_reais.extend(labels.cpu().numpy())

acuracia = accuracy_score(rotulos_reais, previsoes_finais)


print(f"Acurácia Global: {acuracia * 100:.2f}%\n")
print(classification_report(rotulos_reais, previsoes_finais, target_names=['Não Depressivo', 'Depressivo']))

cm = confusion_matrix(rotulos_reais, previsoes_finais)
cor_bert1 = "#76558F"
cores_bert1 = ["#F4EFF7", "#DCCDE5", "#B49BC5", cor_bert1]
cmap_bert1 = LinearSegmentedColormap.from_list("BERT1", cores_bert1)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap=cmap_bert1,
    xticklabels=['Não Depressivo', 'Depressivo'],
    yticklabels=['Não Depressivo', 'Depressivo'],
    linewidths=1,
    linecolor='white'
)
plt.xlabel('Classe prevista', fontsize=11)
plt.ylabel('Classe real', fontsize=11)
plt.title('Matriz de Confusão ', fontsize=13, fontweight='bold', pad=15)
plt.tight_layout()
plt.show()
```

```python
# Curva ROC e AUC
probabilidades, verdadeiros = [], []

meu_modelo.eval()
with torch.no_grad():
    for batch in test_loader:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels = batch['labels'].to(device)

        logits = meu_modelo(input_ids, attention_mask)
        probs = F.softmax(logits, dim=1)

        probabilidades.extend(probs[:, 1].cpu().numpy())
        verdadeiros.extend(labels.cpu().numpy())

fpr, tpr, limiares = roc_curve(verdadeiros, probabilidades)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(7, 5))
plt.plot(fpr, tpr, color="#76558F", lw=2.5, label=f'AUC = {roc_auc:.4f}')
plt.fill_between(fpr, tpr, alpha=0.08, color="#76558F")
plt.plot([0, 1], [0, 1], color="#C9C0CF", lw=2, linestyle='--', label='Classificador aleatório')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('Taxa de Falsos Positivos', fontsize=11)
plt.ylabel('Taxa de Verdadeiros Positivos', fontsize=11)
plt.title('Curva ROC ', fontsize=13, fontweight='bold', pad=15)
plt.legend(loc="lower right", frameon=True)
plt.grid(alpha=0.25, linestyle='--')
plt.tight_layout()
plt.show()

print(f"AUC Final: {roc_auc:.4f}")
```

```python
#LIME
def predict_probs_lime(texts):
    meu_modelo.eval()
    todas_probs = []
    for i in range(0, len(texts), 32):
        lote_textos = texts[i:i + 32]
        inputs = tokenizer(lote_textos, padding='max_length', truncation=True, max_length=128, return_tensors="pt")
        input_ids = inputs['input_ids'].to(device)
        attention_mask = inputs['attention_mask'].to(device)

        with torch.no_grad():
            logits = meu_modelo(input_ids, attention_mask)
            probs = F.softmax(logits, dim=1)
            todas_probs.append(probs.cpu().numpy())

    return np.vstack(todas_probs)

explainer = LimeTextExplainer(class_names=['Não Depressivo', 'Depressivo'], random_state=42)

idx = 72
texto_amostra = str(df_teste.iloc[idx]['texto_limpo'])
rotulo_real = int(df_teste.iloc[idx]['label'])

print(f"Texto analisado (amostra #{idx}):\n{texto_amostra}\n")
print(f"Rótulo real: {rotulo_real} ({'Depressivo' if rotulo_real == 1 else 'Não Depressivo'})\n")

exp = explainer.explain_instance(texto_amostra, predict_probs_lime, num_features=10, num_samples=500)
lista_features = exp.as_list()
palavras = [item[0] for item in lista_features]
pesos = np.array([item[1] for item in lista_features])

plt.figure(figsize=(8, 5))
cores = ['#76558F' if peso > 0 else '#557A68' for peso in pesos]
plt.barh(palavras[::-1], pesos[::-1], color=cores[::-1])
plt.axvline(0, color='#B8B8B8', linewidth=1.2)
plt.xlabel('Impacto na decisão do modelo', fontsize=11)
plt.ylabel('Palavras (features)', fontsize=11)
plt.title(f'Explicação LIME (Amostra #{idx})', fontsize=13, fontweight='bold', pad=15)
plt.grid(axis='x', linestyle='--', alpha=0.25)
plt.tight_layout()
plt.show()
```
